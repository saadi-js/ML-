```python
import pandas as pd
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Embedding,SimpleRNN, Dense
```

```python
import numpy as np

sentences = [
 "I love this product",
 "This movie made me smile",
 "Service was friendly and quick",
 "Today felt bright and happy",
 "This is the best day",
 "Absolutely fantastic experience",
 "I enjoyed every single moment",
 "Great job, well done",
 "The food tasted delicious",
 "Totally recommend to everyone",
 "Very satisfied with results",
 "This worked better than expected",
 "Amazing quality and value",
 "Such a pleasant surprise",
 "I feel positive about this",
 "I hate this product",
 "This movie bored me",
 "Service was rude and slow",
 "Today was cold and lonely",
 "This is the worst day",
 "Terrible experience overall",
 "I regret buying this",
 "Very disappointed with results",
 "The food tasted awful",
 "Do not recommend this",
 "It broke after one use",
 "Not worth the money",
 "Utterly frustrating and annoying",
 "I feel negative about this",
 "Such a waste of time",
]
labels = [1]*15 + [0]*15
labels = np.array(labels)
     
```

```python
vocab_size=2000
tokenizer=Tokenizer(num_words=vocab_size,oov_token="<OOV>")
tokenizer.fit_on_texts(sentences)
seqs=tokenizer.texts_to_sequences(sentences)
max_len=max([len(seq) for seq in seqs])
X=pad_sequences(seqs,maxlen=max_len,padding="post")
Y=labels
```

```python
embed_dim=16
rnn_units=8

```

```python

inp = Input(
    shape=(max_len,),
    dtype="int32",
    name="input"
)

x = Embedding(
    input_dim=vocab_size,
    output_dim=embed_dim,
    mask_zero=True,
    name="embed"
)(inp)

rnn = SimpleRNN(
    units=rnn_units,
    name="simple_rnn",
    return_sequences=False,
    return_state=False
)(x)

out = Dense(
    1,
    activation="sigmoid",
    name="output"
)(rnn)

model = Model(
    inputs=inp,
    outputs=out
)

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
model.summary()
```

```python
model.fit(
    X,
    Y,
    epochs=25,
    batch_size=8,
    
    verbose=1
)
```

```python
import numpy as np

sequence_rnn = SimpleRNN(
    units=rnn_units,
    return_sequences=True,
    name="sequence_rnn"
)

hidden_sequence = sequence_rnn(x)

sequence_rnn.set_weights(
    model.get_layer("simple_rnn").get_weights()
)

intermediate_model = Model(
    inputs=model.input,
    outputs=hidden_sequence,
    name="intermediate_model"
)

hidden_values = intermediate_model.predict(X, verbose=0)

sample_index = 0
sample_tokens = X[sample_index]
sample_hidden_states = hidden_values[sample_index]

np.set_printoptions(precision=3, suppress=True)

print("Copied RNN weights into sequence-inspection RNN.")
print("Sentence:", sentences[sample_index])
print("Token IDs:", sample_tokens)
print(
    "Hidden states per timestep shape:",
    sample_hidden_states.shape
)
print("Hidden states (timesteps x units):")
print(sample_hidden_states)
```
